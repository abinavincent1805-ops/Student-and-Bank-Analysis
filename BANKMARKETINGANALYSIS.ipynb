## 8. Loading the Dataset

The Bank Marketing dataset was loaded from the ZIP file using Python's ZipFile module and Pandas.

The file bank-full.csv was selected because it contains the complete dataset with 45,211 customer records and 17 attributes.

```python
with zipfile.ZipFile("bank+marketing.zip", "r") as outer_zip:
    bank_data = outer_zip.read("bank.zip")

with zipfile.ZipFile(BytesIO(bank_data), "r") as inner_zip:
    with inner_zip.open("bank-full.csv") as csv_file:
        df = pd.read_csv(csv_file, sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## 14. Term Deposit Subscription Analysis

The target variable 'y' was analysed to understand the number of customers who subscribed to the term deposit and those who did not.

The value_counts() function was used to calculate the number of customers in each category.

The subscription percentages were also calculated to understand the distribution of the target variable.

```python
print("TERM DEPOSIT SUBSCRIPTION COUNT")
print(df["y"].value_counts())

print("\nSubscription percentage:")
print(df["y"].value_counts(normalize=True) * 100)
```

## 15. Subscription Distribution Visualization

A count plot was created to visualize the distribution of customers based on term deposit subscription.

The graph compares:
- Customers who subscribed to the term deposit (Yes).
- Customers who did not subscribe (No).

This visualization helps us understand the balance between the two categories.

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="y", data=df)

plt.title("Term Deposit Subscription Distribution")
plt.xlabel("Subscription")
plt.ylabel("Number of Customers")

plt.show()
```

## 16. Customer Age Analysis

The average age of customers was calculated based on their term deposit subscription status.

The groupby() and mean() functions were used to compare the average age of customers who subscribed and those who did not.

A bar chart was also created to visualize the comparison.

This analysis helps us understand the relationship between customer age and term deposit subscription.

```python
age_subscription = df.groupby("y")["age"].mean()

print("Average Customer Age Based on Subscription:")
print(age_subscription)
```

```python
age_subscription.plot(kind="bar", figsize=(7, 5), edgecolor="black")

plt.title("Average Customer Age Based on Subscription")
plt.xlabel("Term Deposit Subscription")
plt.ylabel("Average Customer Age")
plt.xticks(rotation=0)
plt.show()
```

## 17. Job Type Analysis

The subscription rate was calculated for different job categories.

Customers were grouped according to their job type, and the percentage of customers who subscribed to the term deposit was calculated.

A bar chart was created to compare subscription rates across different occupations.

This analysis helps identify how subscription behaviour varies among different job categories.

```python
job_subscription = df.groupby("job")["y"].apply(
    lambda x: (x == "yes").mean() * 100
)

print("Subscription Rate Based on Job Type:")
print(job_subscription.sort_values(ascending=False))
```

```python
job_subscription.sort_values().plot(
    kind="bar",
    figsize=(10, 5),
    edgecolor="black"
)

plt.title("Term Deposit Subscription Rate by Job Type")
plt.xlabel("Job Type")
plt.ylabel("Subscription Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

## 25. Correlation Analysis

Correlation analysis was performed to understand the relationships between numerical variables in the dataset.

The corr() function was used to calculate the correlation coefficients.

A heatmap was created using Seaborn to visualize these relationships.

- Values close to +1 indicate a strong positive correlation.
- Values close to -1 indicate a strong negative correlation.
- Values close to 0 indicate a weak linear correlation.

This analysis helps identify relationships between numerical attributes in the Bank Marketing dataset.

```python
plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Bank Marketing Dataset")
plt.show()
```
