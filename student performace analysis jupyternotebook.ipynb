## 3. Loading the Dataset

The Student Performance dataset is loaded using the Pandas library.

The dataset contains information about students, including their demographic details, study habits, school-related factors, and academic grades.

```python
import pandas as pd
import zipfile
from io import BytesIO

with zipfile.ZipFile("student+performance.zip", "r") as outer_zip:
    inner_data = outer_zip.read("student.zip")

with zipfile.ZipFile(BytesIO(inner_data), "r") as inner_zip:
    with inner_zip.open("student-mat.csv") as csv_file:
        df = pd.read_csv(csv_file, sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
df.head()
```

## 5. Loading the Student Performance Dataset

The dataset is stored inside nested ZIP files. Python's `zipfile` library is used to access the files, and Pandas is used to load the `student-mat.csv` dataset into a DataFrame.

The dataset is then checked to confirm that it has been loaded successfully and to identify its dimensions.

```python
import zipfile
import pandas as pd
from io import BytesIO

with zipfile.ZipFile("student+performance.zip", "r") as outer_zip:
    inner_data = outer_zip.read("student.zip")

with zipfile.ZipFile(BytesIO(inner_data), "r") as inner_zip:
    with inner_zip.open("student-mat.csv") as csv_file:
        df = pd.read_csv(csv_file, sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## 7. Checking Missing Values

Missing values are empty or unavailable entries in a dataset.

The `isnull().sum()` function identifies and counts the missing values in each column. This helps us understand the completeness and quality of the dataset.

```python
print("MISSING VALUES IN EACH COLUMN")

print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

## 8. Checking Duplicate Records

Duplicate records are repeated entries in a dataset.

The `duplicated().sum()` function counts the total number of duplicate rows. This helps us identify repeated student records and maintain data quality.

```python
print("Total duplicate rows:")
print(df.duplicated().sum())
```

## 9. Final Grade Analysis

This section analyses the final grades (G3) of students using statistical measures.

- Average: Calculates the mean final grade.
- Highest: Identifies the maximum final grade.
- Lowest: Identifies the minimum final grade.
- Median: Finds the middle value of the final grades.

These measures help us understand the overall academic performance of students.

```python
print("FINAL GRADE ANALYSIS")

print("Average final grade:", df["G3"].mean())
print("Highest final grade:", df["G3"].max())
print("Lowest final grade:", df["G3"].min())
print("Median final grade:", df["G3"].median())
```

```python
import matplotlib.pyplot as plt
```

## 10. Distribution of Final Grades

A histogram is used to visualize the distribution of students' final grades (G3).

This graph helps us understand how final grades are spread across students and identify the most common grade ranges.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

## 11. Study Time Analysis

This section examines the relationship between students' weekly study time and their average final grades.

The `groupby()` function groups students according to their study-time categories, and `mean()` calculates the average final grade (G3) for each category.

This helps us understand how average academic performance varies with weekly study time.

```python
study_grade = df.groupby("studytime")["G3"].mean()

print(study_grade)
```

## 12. Visualization of Study Time and Final Grades

A bar chart is used to compare the average final grades of students across different weekly study-time categories.

This visualization helps identify how academic performance varies among students with different amounts of study time.

```python
study_grade.plot(kind="bar", figsize=(8, 5), edgecolor="black")

plt.title("Average Final Grade Based on Study Time")
plt.xlabel("Study Time")
plt.ylabel("Average Final Grade")
plt.xticks(rotation=0)
plt.show()
```

## 34. Pass and Fail Analysis

This analysis classifies students as Pass or Fail based on their final grades (G3).

Students scoring 10 or above are classified as Pass, while students scoring below 10 are classified as Fail.

The value_counts() function is used to calculate the number of students in each category.

```python
df["Result"] = df["G3"].apply(lambda x: "Pass" if x >= 10 else "Fail")

result_count = df["Result"].value_counts()

print("Pass and Fail Distribution:")
print(result_count)
```

## 35. Visualization of Pass and Fail Distribution

A pie chart is used to represent the percentage of students who passed and failed based on their final grades.

The autopct parameter displays the percentage of each category, making it easier to understand the overall result distribution.

```python
result_count.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    startangle=90
)

plt.title("Pass and Fail Distribution of Students")
plt.ylabel("")
plt.show()
```
